# 10 · Compact synchronized publication export

Export each available modality independently. Missing Eyeball, DriverVideo, or IDUN inputs do not block the others.

Apply accepted affine clock mappings to all finite native timestamps, including extrapolation beyond their observed/fitted support. Support bounds and available support counts remain QC information. No new mapping is fitted. Rerun this notebook before Step 11 to replace earlier CSVs whose derived timestamps were blank outside support.

```python
import importlib
import sys
from pathlib import Path

import pandas as pd
from IPython.display import display

for PROJECT_ROOT in (Path.cwd().resolve(), *Path.cwd().resolve().parents):
    if (PROJECT_ROOT / 'helpers').is_dir():
        sys.path.insert(0, str(PROJECT_ROOT))
        break
else:
    raise FileNotFoundError('Could not locate the BIGPROJECT/helpers folder')

from helpers.project_config import load_project_config
from helpers.sync_quality import driver_lsl_alignment as driver_lsl
from helpers.sync_quality import driver_timestamp_export as driver_export
from helpers.sync_quality import eyeball_time_table as eyeball_export
from helpers.sync_quality import idun_lsl_alignment as idun_export
importlib.reload(driver_lsl)
importlib.reload(driver_export)
importlib.reload(eyeball_export)
importlib.reload(idun_export)

CONFIG = load_project_config(PROJECT_ROOT / 'ImBrAid' / 'config.json')
DRIVER_LANDMARK_ROOT = CONFIG.path('paths', 'driver_landmark_root')
PUBLICATION_ROOT = CONFIG.path('paths', 'publication_root')
EYEBALL_OUTPUT_DIR = PUBLICATION_ROOT / 'Eyeball'
IDUN_OUTPUT_DIR = PUBLICATION_ROOT / 'IDUN'
DRIVER_OUTPUT_DIR = PUBLICATION_ROOT / 'DriverVideo'
ET_LSL_CSV = CONFIG.path('files', 'et_lsl_alignment_csv')
ET_EYEBALL_CSV = CONFIG.path('files', 'et_eyeball_alignment_csv')
DRIVER_ET_CSV = CONFIG.path('files', 'driver_et_alignment_csv')
IDUN_LSL_CSV = CONFIG.path('files', 'idun_lsl_alignment_csv')
EYEBALL_SUMMARY_CSV = CONFIG.path('files', 'eyeball_export_summary_csv')
DRIVER_SUMMARY_CSV = CONFIG.path('files', 'driver_video_export_summary_csv')
IDUN_SUMMARY_CSV = CONFIG.path('files', 'idun_export_summary_csv')
IDUN_ISSUES_CSV = CONFIG.path('files', 'idun_export_issues_csv')
OVERWRITE = True
```

## Available modalities

```python
print('Each available modality will be exported independently.')
```

## Eyeball and DriverVideo publication files

```python
eyeball_summary = None
if ET_EYEBALL_CSV.is_file() and ET_LSL_CSV.is_file():
    et_eyeball = pd.read_csv(ET_EYEBALL_CSV, dtype=str, keep_default_na=False)
    et_lsl = pd.read_csv(ET_LSL_CSV, dtype=str, keep_default_na=False)
    eyeball_summary = eyeball_export.export_aligned_eyeball_csvs(
        et_eyeball, et_lsl, output_dir=EYEBALL_OUTPUT_DIR,
        overwrite=OVERWRITE,
    )
    eyeball_export.save_eyeball_timestamp_summary(
        eyeball_summary, EYEBALL_SUMMARY_CSV,
    )
    display(eyeball_summary)
else:
    print('Eyeball skipped: its mapping files are absent.')

driver_summary = None
if DRIVER_ET_CSV.is_file() and ET_LSL_CSV.is_file() and DRIVER_LANDMARK_ROOT.is_dir():
    driver_summary = driver_export.export_publication_from_synchronized_landmarks(
        landmark_root=DRIVER_LANDMARK_ROOT,
        driver_et_alignment_csv=DRIVER_ET_CSV,
        et_lsl_alignment_csv=ET_LSL_CSV, output_root=DRIVER_OUTPUT_DIR,
        overwrite=OVERWRITE,
        project=CONFIG.project,
    )
    driver_export.save_driver_timestamp_summary(
        driver_summary, DRIVER_SUMMARY_CSV,
    )
    display(driver_summary)
else:
    print('DriverVideo skipped: its landmarks or mapping files are absent.')
```

## IDUN publication files

```python
idun_summary = None
if IDUN_LSL_CSV.is_file():
    idun_mappings = pd.read_csv(IDUN_LSL_CSV, dtype=str, keep_default_na=False)
    idun_summary, idun_issues = idun_export.export_synchronized_idun_data(
        idun_mappings, output_dir=IDUN_OUTPUT_DIR,
        et_lsl_alignment_csv=ET_LSL_CSV if ET_LSL_CSV.is_file() else None,
        overwrite=OVERWRITE,
    )
    idun_export.save_idun_export_summary(idun_summary, IDUN_SUMMARY_CSV)
    idun_export.save_idun_export_issues(idun_issues, IDUN_ISSUES_CSV)
    display(idun_summary)
    display(idun_issues)
else:
    print('IDUN skipped: its mapping file is absent.')
```
