# 00 — Extract DriverVideo landmarks

The Face and Pose task bundles must already exist at the configured model paths. Existing landmark CSVs are skipped whenever `OVERWRITE=False`; set it to `True` to regenerate them.

```python
from pathlib import Path
import sys

START = Path.cwd().resolve()
CANDIDATES = [START]
CANDIDATES.extend(START.parents)
for PROJECT_ROOT in CANDIDATES:
    if (PROJECT_ROOT / 'helpers').is_dir():
        break
else:
    raise RuntimeError('Run this notebook from the BIGPROJECT repository.')
sys.path.insert(0, str(PROJECT_ROOT))

from helpers.DriverVideo import driver_landmarks as dl
from helpers.project_config import load_project_config

CONFIG = load_project_config(PROJECT_ROOT / 'ImBrAid' / 'config.json')
DATA_ROOT = CONFIG.path('paths', 'processing_data_root')
OUTPUT_ROOT = CONFIG.path('paths', 'driver_landmark_root')
DRIVER_VIDEO_SUBFOLDER = CONFIG.value('modalities', 'driver_video_source_subfolder')
FACE_MODEL_PATH = OUTPUT_ROOT / 'models' / 'face_landmarker.task'
POSE_MODEL_PATH = OUTPUT_ROOT / 'models' / 'pose_landmarker_full.task'
FACE_MODEL_URL = "https://storage.googleapis.com/mediapipe-models/face_landmarker/face_landmarker/float16/1/face_landmarker.task"
POSE_MODEL_URL = "https://storage.googleapis.com/mediapipe-models/pose_landmarker/pose_landmarker_full/float16/1/pose_landmarker_full.task"
OVERWRITE = False

LANDMARK_CONFIG = dl.DriverLandmarkerConfig(
    face_roi=(0.25, 0.35, 0.75, 1.00),
    pose_roi=(0.18, 0.28, 0.82, 1.00),
)
```

```python
summary_df = dl.process_all_driver_videos(
    data_root=DATA_ROOT, output_root=OUTPUT_ROOT,
    face_model_path=FACE_MODEL_PATH, pose_model_path=POSE_MODEL_PATH,
    driver_video_subfolder=DRIVER_VIDEO_SUBFOLDER,
    overwrite=OVERWRITE, config=LANDMARK_CONFIG,
    face_model_url=FACE_MODEL_URL, pose_model_url=POSE_MODEL_URL
)
summary_df
```
